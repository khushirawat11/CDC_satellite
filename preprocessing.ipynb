IMPORTS

```python
import numpy as np
import pandas as pd


import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path
from PIL import Image

import geopandas as gpd
import contextily as ctx

plt.style.use("seaborn-v0_8")
sns.set_context("talk")
```

PATH SETUP

```python
PROJECT_ROOT = Path(".").resolve()

DATA_DIR = PROJECT_ROOT / "data"
IMAGE_DIR = DATA_DIR / "images"
PROCESSED_DIR = DATA_DIR / "processed"

PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

print("Images directory:", IMAGE_DIR)
```

LOAD TRAINING DATA

```python
train_df = pd.read_csv("train(1)(train(1)).csv")

print("Train shape:", train_df.shape)
train_df.head()
```

```python
train_df.shape
```

DEFINE KEY COLUMNS

```python
TARGET_COL = "price"
LAT_COL = "lat"
LON_COL = "long"
```

CREATE IMAGE PATH COLUMN

```python
train_df["image_path"] = train_df.index.map(
    lambda x: IMAGE_DIR / f"{x}.png"
)

train_df[["lat", "long", "image_path"]].head()
```

VERIFY IMAGE COVERAGE

```python
train_df["image_exists"] = train_df["image_path"].apply(lambda p: p.exists())

coverage = train_df["image_exists"].mean()
print(f"Image coverage: {coverage:.2%}")

```

LOG-PRICE TRANSFORMATION

```python
train_df["log_price"] = np.log1p(train_df[TARGET_COL])

train_df[[TARGET_COL, "log_price"]].describe().T

```

```python
train_df.isna().mean().sort_values(ascending=False).head(10)
```

```python
train_df[[LAT_COL, LON_COL]].describe()
```

TABULAR EDA

```python
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
sns.histplot(train_df[TARGET_COL], bins=50)
plt.title("Price distribution")

plt.subplot(1,2,2)
sns.histplot(train_df["log_price"], bins=50)
plt.title("Log(price + 1)")

plt.tight_layout()
plt.show()
```

GEOSPATIAL EDA

```python
gdf = gpd.GeoDataFrame(
    train_df.copy(),
    geometry=gpd.points_from_xy(train_df[LON_COL], train_df[LAT_COL]),
    crs="EPSG:4326",
)

gdf_web = gdf.to_crs(epsg=3857)

sample = gdf_web.sample(min(8000, len(gdf_web)), random_state=42)

fig, ax = plt.subplots(figsize=(10,10))
sample.plot(
    ax=ax,
    column="log_price",
    cmap="viridis",
    markersize=5,
    legend=True
)

ctx.add_basemap(ax, source=ctx.providers.CartoDB.Positron)
ax.axis("off")
plt.show()
```

VISUAL EDA

```python
def show_random_images(n=9):
    sample = train_df[train_df["image_exists"]].sample(n, random_state=42)
    fig, axes = plt.subplots(3, 3, figsize=(9,9))
    axes = axes.flatten()

    for ax, (_, row) in zip(axes, sample.iterrows()):
        img = Image.open(row["image_path"]).convert("RGB")
        ax.imshow(img)
        ax.set_title(f"Price: {row[TARGET_COL]:,.0f}")
        ax.axis("off")

    plt.tight_layout()
    plt.show()

show_random_images()
```

CREATE SPATIAL BINS

```python
grid_size = 0.02

lat_bin = (train_df[LAT_COL] / grid_size).round().astype(int)
lon_bin = (train_df[LON_COL] / grid_size).round().astype(int)

train_df["spatial_bin"] = lat_bin.astype(str) + "_" + lon_bin.astype(str)

train_df["spatial_bin"].nunique()
```

TRAIN / VAL SPLIT

```python
from sklearn.model_selection import GroupShuffleSplit

gss = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)

train_idx, val_idx = next(
    gss.split(train_df, groups=train_df["spatial_bin"])
)

train_split = train_df.iloc[train_idx].copy()
val_split = train_df.iloc[val_idx].copy()

print("Train:", train_split.shape)
print("Val:", val_split.shape)
```

```python
train_split = train_split[train_split["image_exists"]]
val_split = val_split[val_split["image_exists"]]

print("Final Train:", train_split.shape)
print("Final Val:", val_split.shape)
```

SAVE FINAL DATA

```python
train_split.to_csv(PROCESSED_DIR / "train_final.csv", index=False)
val_split.to_csv(PROCESSED_DIR / "val_final.csv", index=False)

print("Preprocessing complete ✅ (saved as CSV)")
```

```python
def find_outliers_iqr(df, col):
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]
    return outliers, lower, upper
```

```python
price_outliers, low, high = find_outliers_iqr(train_df, "price")

print("Lower bound:", low)
print("Upper bound:", high)
print("Number of price outliers:", price_outliers.shape[0])

price_outliers[["id", "price"]].head()
```

```python
for col in ["price", "sqft_living", "sqft_lot"]:
    outliers, low, high = find_outliers_iqr(train_df, col)
    print(f"{col}: {outliers.shape[0]} outliers")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load training data
train_df = pd.read_csv("data/processed/train_final.csv")

# Create log_price if not already present
if "log_price" not in train_df.columns:
    train_df["log_price"] = np.log1p(train_df["price"])

# Plot PDFs
plt.figure(figsize=(14, 5))

# ---- RAW PRICE PDF ----
plt.subplot(1, 2, 1)
sns.kdeplot(train_df["price"], fill=True)
plt.title("PDF of House Prices (Raw)")
plt.xlabel("Price")
plt.ylabel("Density")

# ---- LOG PRICE PDF ----
plt.subplot(1, 2, 2)
sns.kdeplot(train_df["log_price"], fill=True)
plt.title("PDF of House Prices (After Log Transformation)")
plt.xlabel("log(price + 1)")
plt.ylabel("Density")

plt.tight_layout()
plt.show()

```

```python
plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
sns.histplot(train_df["price"], bins=50, kde=True)
plt.title("Histogram + PDF (Raw Price)")

plt.subplot(1, 2, 2)
sns.histplot(train_df["log_price"], bins=50, kde=True)
plt.title("Histogram + PDF (Log Price)")

plt.tight_layout()
plt.show()
```
